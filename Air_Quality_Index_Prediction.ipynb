**Handling the missing values**

Convert all the -200 to NaN

Replace all the NaN values with mean of that specific column

```python
air_quality_data = air_quality_data.replace(to_replace = -200, value = np.nan)
```

```python
air_quality_data.isnull().sum()
```

This shows the actual number of missing values

```python
air_quality_data.tail()
```

```python
air_quality_data.mean(numeric_only=True)
```

```python
# replacing the missing values with mean value
air_quality_data = air_quality_data.fillna(air_quality_data.mean(numeric_only=True))
```

**Forecasting with FB Prophet Algorithm**

FB Prophet Documentation: https://facebook.github.io/prophet/docs/quick_start.html

```python
# converting the date format from DD/MM/YYYY to YYYY-MM-DD

date_info = pd.to_datetime(air_quality_data['Date'], dayfirst=True)
print(date_info)
```

```python
time_info = air_quality_data['Time']
print(time_info)
```

```python
time_info = time_info.apply(lambda x: x.replace('.',':'))
```

```python
print(time_info)
```

```python
print(type(date_info))
print(type(time_info))
```

```python
# combining two series to a pandas dataframe

date_time = pd.concat([date_info, time_info], axis=1)
```

```python
date_time.head()
```

```python
date_time.shape
```

```python
# combining date and time columns as one
date_time['ds'] = date_time['Date'].astype(str) + ' ' + date_time['Time'].astype(str)
```

```python
date_time.head()
```

```python
date_time.info()
```

We have to convert ds from object datatype to date time format

```python
data = pd.DataFrame()
```

```python
data['ds'] = pd.to_datetime(date_time['ds'])
```

```python
data.head()
```

```python
data['y'] = air_quality_data['RH']
```

```python
data.head()
```

```python
!pip install prophet
```

```python
from prophet import Prophet
```

```python
# training the Prophet model
model = Prophet()
model.fit(data)
```

```python
future = model.make_future_dataframe(periods=365, freq='H')
future.tail()
```

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
fig1 = model.plot(forecast)
```

```python
fig2 = model.plot_components(forecast)
```
